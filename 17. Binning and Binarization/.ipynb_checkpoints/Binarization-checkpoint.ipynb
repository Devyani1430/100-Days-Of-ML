```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer

from sklearn.metrics import accuracy_score

from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','SibSp','Parch','Survived'])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df['Family'] = df['SibSp']+df['Parch']
```

```python
df.head()
```

```python
df.drop(columns=['SibSp','Parch'],inplace=True)
```

```python
df.head()
```

```python
X = df.drop(columns=['Survived'])
Y = df['Survived']
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

## Without Binarization

```python
clf=DecisionTreeClassifier()

clf.fit(x_train,y_train)

y_pred=clf.predict(x_test)

accuracy_score(y_pred,y_test)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),X,Y,cv=10,scoring="accuracy"))
```

## Applying Binarization

```python
from sklearn.preprocessing import Binarizer
```

```python
trf=ColumnTransformer([
    ("bin",Binarizer(copy=False),["Family"])
],remainder="passthrough")

# We want changes in existing column that's why we kept 'copy = False' in the above function
```

```python
X_train_trf=trf.fit_transform(x_train)
X_test_trf=trf.transform(x_test)
```

```python
pd.DataFrame(X_train_trf,columns=["Family","Age","Fare"])
```

```python
clf1=DecisionTreeClassifier()

clf1.fit(X_train_trf,y_train)

Y_pred2=clf1.predict(X_test_trf)

accuracy_score(Y_pred2,y_test)
```

```python
X_trf=trf.transform(X)

np.mean(cross_val_score(DecisionTreeClassifier(),X_trf,Y,cv=10,scoring="accuracy"))
```
