## Working with Dates

```python
# Converting to DATE-TIME datatype

date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

#### 1. Extracting year

```python
date['date_year'] = date['date'].dt.year
```

```python
date.head()
```

#### 2. Extracting Month

```python
date['date_month'] = date['date'].dt.month
```

```python
date.sample(5)
```

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date.sample(5)
```

#### 3. Extracting days

```python
date['date_day'] = date['date'].dt.day
```

```python
date.sample(5)
```

```python
# Day of the week

date['date_day_of_week'] = date['date'].dt.dayofweek
```

```python
date.sample(5)
```

```python
date['date_day_name'] = date['date'].dt.day_name()
date.sample(5)
```

```python
# is it weekend..?
date['date_is_weekend'] = np.where(date['date_day_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

#### 4. Extract Week of the year

```python
date['date_week'] = date['date'].dt.weekday

date.drop(columns=['product_id','city_id','orders']).head()
```

### 5. Extract Quarter

```python
date["date_quarter"] = date["date"].dt.quarter
date.drop(columns=["product_id","orders","city_id"]).head()
```

#### 6. Extract Semester

```python
date["semester"] = np.where(date["date_quarter"].isin([1,2]),1,2)

date.drop(columns=['product_id','city_id','orders']).head()
```

#### 7. Extract time elapsed between dates

```python
import datetime

today=datetime.datetime.today()
```

```python
today
```

```python
today-date["date"]
```

```python
(today-date["date"]).dt.days
```

```python
# Days Passed

np.round((today - date['date']) / np.timedelta64(1, 'D') / 30.44, 0)
```

### Working with time

```python
time
```

```python
# Converting to datetime datatype

time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

#### Extract Time Part

```python
time['time'] = time['date'].dt.time

time.head()
```

#### Extract Time Diffrence

```python
today-time["date"]
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# in minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# in hours

(today - time['date'])/np.timedelta64(1,'h')
```
