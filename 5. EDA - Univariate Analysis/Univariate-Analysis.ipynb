```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

# Categorical Data    
let's 1st focus on categorical data

### Countplot

```python
# sns.countplot(df['Survived'])  ....... this is older syntax may give diffrent result or error in new syntax u need to define axis
sns.countplot(x = df['Survived'])

df['Survived'].value_counts()
# '.value_counts()' gives total number of each elements in the colmn



# Conclusion = here u can see no of people survived is less than people who didn't survive
```

```python
ax = df["Sex"].value_counts().plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container)

plt.show()

# Conclusion = Male are more than Female 
```

```python
sns.countplot(x=df['Pclass'],hue=df['Pclass'],legend=False)
df['Pclass'].value_counts()
```

```python
sns.countplot(x = df['Embarked'])
df['Embarked'].value_counts()
```

### Pie Chart

```python
df['Survived'].value_counts().plot(kind = 'pie' , autopct = '%.2f')
# 'autopct' used to display percentage

# Conclusion = 61% dead and 38% survived
```

```python
df['Pclass'].value_counts().plot(kind = 'pie', autopct = "%.2f")
```

# Numerical Data

### a. Histogram

```python
plt.hist(df['Age'])
```

```python
plt.hist(df['Age'],bins=40)
```

```python
plt.hist(df['Age'],bins=5)

# u can change the range using bins...5 means 5 intervals , 40 means 40 intervals, 10 is default maybe
```

### b. Distplot

```python
sns.displot(df['Age'],kde = True,color='green',alpha = 0.3,edgecolor = "none")

# 'kde' : shows the curve line 
# colour : change color 
# alpha : can manage translucency of bars  
# edgecolor : removed the outline of the bars
```

```python
sns.boxplot(df['Age'])
```
