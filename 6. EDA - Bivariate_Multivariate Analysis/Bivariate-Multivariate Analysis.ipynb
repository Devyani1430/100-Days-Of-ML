# 2 . Numerical - Catagorical

### a. BarPlot

```python
sns.barplot(data = Titanic, x='Pclass',y='Age', hue = 'Pclass')

# tip = usually  we put categorical column on x-axis and numerical column on y-axis
```

```python
sns.barplot(data = Titanic, x='Pclass',y='Fare', hue = 'Pclass')
```

```python
sns.barplot(data = Titanic, x='Pclass',y='Fare', hue = 'Sex')

# Multivariate analysis
```

### b. BoxPlot

```python
sns.boxplot(x = Titanic['Sex'], y = Titanic['Age'])
```

```python
sns.boxplot(x = Titanic['Sex'], y = Titanic['Age'],hue = Titanic['Survived'])
```

### c. DistPlot

```python
sns.distplot(Titanic['Age'])
```

```python
sns.distplot(Titanic[Titanic["Survived"]==0]["Age"] ,hist=False)     # Those who didn't survive
sns.distplot(Titanic[Titanic["Survived"]==1]["Age"] ,hist=False)     # Those who survived

# if u notice u can see if u r too young then ur probablity of surviving is more than dying

# u Just have notice observe and perform EDA properly to understand the data
```

### Categorical - Categorical

### a. Heatmap

```python
Titanic.head()
```

```python
pd.crosstab(Titanic['Pclass'], Titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(Titanic['Pclass'], Titanic['Survived']))
```

```python
(Titanic.groupby('Pclass').mean(numeric_only=True)['Survived']*100).plot(kind = 'bar')
```

```python
pd.crosstab(Titanic['SibSp'],Titanic['Survived'])
```

```python
sns.clustermap(pd.crosstab(Titanic["Parch"],Titanic["Survived"]))
```
